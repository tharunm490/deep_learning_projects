```python
img
# not a nupy array, it is a PIL image object. We need to convert it into a numpy array before we can use it for data augmentation.
```

```python
datagen=ImageDataGenerator(
  rotation_range=30, 
  width_shift_range=0.2,
    height_shift_range=0.2, 
    shear_range=0.2, zoom_range=0.2, 
    horizontal_flip=True, fill_mode='nearest')
```

```python
img=image.img_to_array(img)
```

```python
img.shape
```

```python
input_batch=img.reshape(1,200,200,3)
```

```python
import os
save_dir = '/content/drive/MyDrive/augmented_images'
os.makedirs(save_dir, exist_ok=True)

i=0
for output in datagen.flow(input_batch, batch_size=1, save_to_dir=save_dir):
  i=i+1
  if i==20:
    break
```
